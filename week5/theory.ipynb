## Ex 1: Break Points and Growth Functions 

-   What is the (smallest) break point for the hypothesis set consisting
    of circles centered around $(0,0)$? 
    
    For a given circle the
    hypothesis returns $1$ for points inside the circle and $-1$ for
    points outside. 

    **My intuition is: If I can make any set of points, that a circle cannot shatter, then that set is a break point. This means that for all points, they should be able to be withnin AND outside the circle.**

    **Example: With 3 points denoted 0 for outside and 1 for inside:**
    **For 1 point:** 0, 1 (shattered)
    **For 2 points:** 00, 01, 10, 11 (is this legal?)

    **I would argue that the break point is then 2. I cannot shatter them with a centered circle, as I cannot make a circle that contains the outer point and excludes the inner point.**


    What is the growth function?

    **The growth function $m_H(N)$ asks: What is the maximum number of different labelings that $H$ can produce on $N$ points?**
    **For 1 point: 2 (inside or outside)**
    **for 2 points: 3 (00, 01, 11) - cannot do 10**
    **for 3 points: 4 (000, 001, 011, 111) - cannot do 010, 100, 101**

    **This gives the growth function $m_H(N) = N + 1$ for $N \geq 1$.**



-   What if we move to centered balls in the 3-dimensional space
    ${{\mathbb R}}^3$? Or in general $d$-dimensional space
    ${{\mathbb R}}^d$ (hyperspheres)

    **Exact same reasoning as above. We can shatter 1 point, but we cannot shatter 2 points, as we the n-dimensional sphere is CENTERED and for any set of points, we cannot create a hypersphere that contains one point and excludes another AND vice versa. Only one of them. The break point is still 2. The growth function is still $m_H(N) = N + 1$ for $N \geq 1$.**



-  Show that the growth function for a singleton hypothesis class $H = \{h\}$ is 1


$$
m_H(n) = max_{x_1, \dots, x_n \in X} |h(x_1), \dots, h(x_n) : h \in H| 
$$
**We have $h(x_n) = h$ for all $n$. Therefore the \"Number of distinct ways H can classify the n points.\" would be 1. Therfore the growth function is always $m_H(n) = 1$.**
-   Is there always a break point for a finite hypothesis set of $n$
    hypotheses? If so, can you give a an upper bound? What is the growth
    function?


**Yes. The break point is at most $log_2(n)$, where $n$ is the number of hypotheses.**

**The growth function $m_H(N)$ asks: What is the maximum number of different labelings that $H$ can produce on $N$ points?**

**The growth function would then be upper bounded by $2^N$, of course. A lower upper bound would be $n = |H|$, as per the argumentation from the last exercise.**


-   Does the set of all functions have a break point? What is its growth
    function?

**No. The set of all functions does not have a break point, as it can shatter any finite set of points. In binary classification the growth function for the set of all functions is $m_H(N) = 2^N$, since for each of the $N$ points, there are 2 possible labels (0 or 1), leading to $2^N$ different labelings.**


## Ex 2: VC Dimension 


-   Does VC Dimension depend on the learning algorithm? Does it depend on the actual
    data set given?

It would also not depend on the data set - the data set might contain an amount of points above or below the VC dimension, but the VC dimension for our hypothesis set does not change.**

-   Does VC Dimension depend on the probability distribution generating
    the data (not the labels)?'

    **Nope - just the hypothesis set**

-   What is the VC Dimension for the hypothesis set consisting of
    circles centered around 0?

    **We saw in the previous exercise that the break point was $2$. Therefore the VC-dimension is $2-1=1$**

-   What if we move to balls (3d)? or in general d dimensions

    **You can argue from shattering, saying if $H_1$ shatters some set, then $H_2$ will ALWAYS be able to shatter the same set, and maybe more. Therefore break point rises, and therefore VC Dimension rises.**

-   Can you give an upper bound on the VC dimension of a finite set of
    $M$ hypotheses?

    **It would be the bound for break points-1 so that is $log_2(n)-1$**

-   What is the maximal VC dimension possible of the intersection of
    hypothesis sets $\mathcal{H}_1,\dots,\mathcal{H}_n$ with VC
    dimension $v_1,\dots,v_n$ 

    **Notice that the intersection of the sets, will make a subset of all of them. Actually; the biggest subset that is a subset of all of them. So what is the largest VC dimension?**

    *So $\mathcal{H}_1 \cap \dots \cap \mathcal{H}_n \subseteq H_i$ for all $i$ $\implies$ $VC_{\mathcal{H}_1 \cap \dots \cap \mathcal{H}_n} \leq VC_{H_i}$ for all $i$.**

    **This means that the VC dimension, is upper bounded by the VC dimension of the initial set that is the smallest.**

Consider the input space $\mathcal{X} = \{1\} \times \mathbb{R}^d$ (with the first coordinate being the constant 1). Show that the VC dimension of the hypothesis space $\mathcal{H} = \{\textrm{sign}(w^\intercal x) \mid w\in \mathbb{R}^{d+1} \}$ corresponding to the perceptron is $d+1$.

We need to show 
1. That there exists a data set of size d+1 that can be shattered by hyperplanes
2. That no data set of size d+2 can be shattered by hyperplanes

We will give a few more hints than the book does.
### Shattering d+1 points
As the book hints you must create an "easy" data set that you store in matrix $X$. 

**Hint:** 
<details>
We suggest you consider as a data matrix, the $(d+1) \times (d+1)$ matrix $X$ whose first column is all-1s (required since $\mathcal{X} = \{1\} \times \mathbb{R}^d$) and where the lower $d \times d$ corner of the matrix is the $d \times d$ identity matrix.

Show that you can construct any dichotomy $y \in \{-1,+1\}^{d+1}$ using some $h \in \mathcal{H}$ and the data matrix $X$ defined above. That is, you have to show that for any $y \in \{-1,+1\}^{d+1}$, you can find some hypothesis $w$ such that for all $i$, we have $\textrm{sign}(w^\intercal x_i)=y_i$ where $x_i$ is the $i$'th row of $X$.
</details>


### No Shattering of d+2 points.
Must show that for any d+2 points, there is a  dichotomy hyperplanes can not capture.

**Hint:**
<details>
Consider an arbitrary set of d+2 points $x_1,\dots, x_{d+2}$ of dimension (d+1) and think of them as vectors in $\{1\} \times \mathbb{R}^d \subset \mathbb{R}^{d+1}$.
Since we have more vectors than dimensions the vectors must be linearly dependent.

i.e. there is a $j$ such that:
$$
x_j = \sum_{i\neq j} a_i x_i
$$
Since $x_j$ is determined by the other data points then so is $w^\intercal x_j$ for any $w$. This means the classification on point $x_j$ is dictated by the classification of the other data points and thus cannot freely be chosen.
i.e.
$$
w^\intercal x_j = w^\intercal \sum_{i\neq j} a_i x_i =\sum_{i\neq j} a_i w^\intercal x_i
$$
Define an impossible dichotomy as follows. 
$$
y_i = \textrm{sign}(a_i), \quad i\neq j, \quad y_j = -1
$$
Show this dichotomy is impossible!
